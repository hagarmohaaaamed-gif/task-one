2. Go to the data folder and use the **[countries_details_data.py]()** file. 
   1. What are the total number of languages in the data ( output:  112 )
   2. Find the ten most spoken languages from the data ( output:  English , French , Arabic )
   3. Find the 10 most populated countries in the world ( output: China, India, US )

````python
languages=set()
for country in countries_data:
    for language in country["languages"]:
        languages.add(language)
print(len(languages))
````

````python
from collections import Counter

languages = []

for country in countries_data:
    for language in country['languages']:
        languages.append(language)

language_count = Counter(languages)

print(language_count.most_common(10))
````

````python
top_countries = sorted(
    countries_data,
    key=lambda country: country['population'],
    reverse=True
)

for country in top_countries[:10]:
    print(country['name'], country['population'])
````
